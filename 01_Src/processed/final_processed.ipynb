```python
drop_list =['projectID','backersCount', 'phaseLabel', 'isFeatured', 'installmentMinPayment', 'hasLimitedStock', 'productCanBePurchased']
bool_list= ['enableBoardGameProperties','isDiscounted']

df.drop(columns=drop_list,inplace=True)
df[bool_list] = df[bool_list].astype(int)
```

```python
# minplayers, maxplayers, minage, playtime
df['minPlayers'] = df['minPlayers'].fillna(1)
df['maxPlayers'] = df['maxPlayers'].fillna(4)
df['minAge'] = df['minAge'].fillna(14)
df['playTime'] = df['playTime'].fillna(30)
```

```python
target_cols = df.loc[:, 'creator_id_0':'Shipping / Price / Tax_1'].columns
df[target_cols] = df[target_cols].fillna(0)
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

currency_map = {
    '€': 'EUR',
    '$': 'USD',
    '£': 'GBP',
    'A$': 'AUD',
    'C$': 'CAD',
    'zł': 'PLN'
}
df['currencySymbol'] = df['currencySymbol'].map(currency_map)

try:
    encoder = OneHotEncoder(sparse_output=False, dtype=int)
except TypeError:
    encoder = OneHotEncoder(sparse=False, dtype=int)

encoded_array = encoder.fit_transform(df[['currencySymbol']])

encoded_df = pd.DataFrame(
    encoded_array, 
    columns=encoder.get_feature_names_out(['currencySymbol']),
    index=df.index
)

df = pd.concat([df.drop(columns=['currencySymbol']), encoded_df], axis=1)
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
import pandas as pd

df['projectTags'] = df['projectTags'].apply(lambda x: [tag.strip() for tag in x.split(',') if tag.strip()])

mlb = MultiLabelBinarizer()
tags_encoded = mlb.fit_transform(df['projectTags'])

df_tags = pd.DataFrame(tags_encoded, columns=mlb.classes_)

df = pd.concat([df.drop(columns=['projectTags']), df_tags], axis=1)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
all_unique_colors = pd.concat([
    df['ks_color_1'], 
    df['ks_color_2'], 
    df['ks_color_3'], 
    df['ks_color_4']
]).unique()

le_color = LabelEncoder()
le_color.fit(all_unique_colors)


color_cols = ['ks_color_1', 'ks_color_2', 'ks_color_3', 'ks_color_4']
for col in color_cols:
    df[col] = le_color.transform(df[col])
```

```python
from sklearn.preprocessing import LabelEncoder
all_unique_emotion = pd.concat([
    df['emotion_adjective_1'], 
    df['emotion_adjective_2'], 
    df['emotion_adjective_3'], 
    df['emotion_adjective_4']
]).unique()

le_color = LabelEncoder()
le_color.fit(all_unique_emotion)


color_cols = ['emotion_adjective_1', 'emotion_adjective_2', 'emotion_adjective_3', 'emotion_adjective_4']
for col in color_cols:
    df[col] = le_color.transform(df[col])
```

```python
df.to_csv('../../02_Data/processed/real_final_ml.csv', index=False, encoding="utf-8-sig")
```
