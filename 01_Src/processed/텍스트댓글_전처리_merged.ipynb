### 기준은 3일로!

```python
df_co['created_at'] = pd.to_datetime(df_co['created_at'], format='ISO8601', utc=True, errors='coerce')
df_co['campaignStart'] = pd.to_datetime(df_co['campaignStart'], format='ISO8601', utc=True, errors='coerce')
df_co['campaignEnd'] = pd.to_datetime(df_co['campaignEnd'], format='ISO8601', utc=True, errors='coerce')

# 캠페인 시작 이전 댓글
df_co_before_start = df_co[
    df_co['created_at'] < df_co['campaignStart']
].copy()

# 캠페인 시작 후 3일 댓글
df_co_3 = df_co[
    (df_co['created_at'] >= df_co['campaignStart']) &
    (df_co['created_at'] <= df_co['campaignStart'] + pd.Timedelta(days=3))
].copy()
```

```python
df_co_before_start["group"] = 0 # upcoming 댓글들
```

```python
df_co_3["group"] = 1 # 3일 후 댓글들
```

```python
df_co_before_start.columns.equals(df_co_3.columns)
```

```python
df_merged = pd.concat([df_co_before_start, df_co_3], axis=0, ignore_index=True)
```

### 기간 통합(group컬럼) csv 저장

```python
#df_merged.to_csv('comments_merged.csv', index=False,
#    encoding="utf-8-sig")
```

```python
#df_merged = pd.read_csv('comments_merged.csv', encoding='utf-8-sig')
```

```python
# 원래 행 순서 저장
df_merged = df_merged.reset_index(drop=True)
df_merged["row_order"] = df_merged.index

# 부모댓글+대댓글 묶음 ID => thread_id = parent_id
df_merged["thread_id"] = df_merged["parent_id"].fillna(df_merged["comment_id"]).astype(int)
```

```python
df_merged.columns
```

```python
text_col = "text"  # 댓글 내용 열 이름으로 바꿔줘. 예: "text_kr"

df_merged["comment_with_label"] = df_merged.apply(
    lambda row: str(row[text_col]) # '[부모댓글]' +
    if pd.isna(row["parent_id"])
    else str(row[text_col]),
    axis=1
)
```

```python
df_thread = (
    df_merged
    .sort_values("row_order") 
    .groupby("thread_id", as_index=False, sort=False)
    .agg(
        projectID=("projectID", "first"),
        group=("group", "first"), # 정렬 'row order' => 각 thread_id 안에서 row_order가 가장 빠른 댓글의 group 값이 들어감
        comment_ids=("comment_id", lambda x: list(x)),
        n_comments=("comment_id", "count"),
        merged_comment=("comment_with_label", lambda x: " ".join(x.dropna().astype(str))),

        creator_id=("creator_id", lambda x: x.isin([1, 2]).sum()),      # 1 또는 2인 경우만 1로 체크하여 합산
        is_pledge_master=("is_pledge_master", "sum"),                  # True인 개수 합산
        is_backer=("is_backer", "sum"),                                # True인 개수 합산
        is_prior_backer=("is_prior_backer", "sum"),                    # True인 개수 합산
        is_pathfinder=("is_pathfinder", "sum")
    )
)

df_thread.head(10)
```

```python
text_col = "text"  # 댓글 내용 열 이름

df_merged["comment_with_label"] = df_merged.apply(
    lambda row: str(row[text_col]) if pd.isna(row["parent_id"]) else str(row[text_col]),
    axis=1
)

# 💡 안전하게 수정된 스레드 압축 코드
df_thread = (
    df_merged
    .sort_values("row_order") 
    # 💡 핵심: projectID와 thread_id 두 개를 기준으로 동시에 묶어줍니다.
    .groupby(["projectID", "thread_id"], as_index=False, sort=False)
    .agg(
        # group은 정렬된 상태에서 각 projectID + thread_id 조합 내 첫 값을 가져옴
        group=("group", "first"), 
        comment_ids=("comment_id", lambda x: list(x)),
        n_comments=("comment_id", "count"),
        merged_comment=("comment_with_label", lambda x: " ".join(x.dropna().astype(str))),

        creator_id=("creator_id", lambda x: x.isin([1, 2]).sum()),      
        is_pledge_master=("is_pledge_master", "sum"),                  
        is_backer=("is_backer", "sum"),                                
        is_prior_backer=("is_prior_backer", "sum"),                    
        is_pathfinder=("is_pathfinder", "sum")
    )
)

df_thread.head(10)
```

```python
df_thread["merged_comment"] = (
    df_thread["merged_comment"]
    .astype(str)
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
)
```
