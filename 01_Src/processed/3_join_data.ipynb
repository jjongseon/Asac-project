### 환율 통일(시점 달라지므로 될 수 있으면 실행X 파일 받아서 아래 read_csv로)

```python
from datetime import date, timedelta
import requests
import numpy as np
import pandas as pd
import time

# ==========================================
# 1. 기준 날짜 세팅
# ==========================================
end_date = date.today()
start_date_1m = end_date - timedelta(days=30)
start_date_6m = end_date - timedelta(days=180)

# ==========================================
# 2. 지워진 환율 계산 함수 복구
# ==========================================
def get_avg_rates_to_usd(base_currency, start_1m, start_6m, end_dt):
    """
    한 번의 API 호출로 1개월, 6개월 평균 환율을 동시에 계산하여 반환
    """
    if base_currency == "USD" or pd.isna(base_currency):
        return 1.0, 1.0

    url = (
        f"https://api.frankfurter.app/"
        f"{start_6m}..{end_dt}"
        f"?base={base_currency}&symbols=USD"
    )

    res = requests.get(url)
    if res.status_code != 200:
        return np.nan, np.nan # 에러 발생 시 결측치 반환
        
    data = res.json()
    rates_1m = []
    rates_6m = []
    
    start_1m_str = start_1m.isoformat()

    for day, rate_info in data["rates"].items():
        if "USD" in rate_info:
            val = rate_info["USD"]
            rates_6m.append(val)
            if day >= start_1m_str:
                rates_1m.append(val)

    avg_1m = sum(rates_1m) / len(rates_1m) if rates_1m else np.nan
    avg_6m = sum(rates_6m) / len(rates_6m) if rates_6m else np.nan

    return avg_1m, avg_6m
```

```python
# ==========================================
# 1. 만들어두신 통화 변환 코드 (그대로 사용!)
# ==========================================
symbol_to_code = {
    "$": "USD",
    "€": "EUR",
    "£": "GBP",
    "A$": "AUD",
    "C$": "CAD",
    "S$": "SGD",
    "zł": "PLN",
    "kr": "SEK",
    "kr.": "SEK",
    "Fr.": "CHF",
    "HK$": "HKD",
    "NZ$": "NZD",
    "Mex$": "MXN",
}

# 임시 정리용 컬럼 생성
final_df["_currencySymbol_clean"] = final_df["currencySymbol"].astype("string").str.strip()

# 최종으로 남길 currencyCode 컬럼 추가
final_df["currencyCode"] = final_df["_currencySymbol_clean"].map(symbol_to_code)

# 매핑 안 된 통화기호 확인
unmapped = final_df[final_df["currencyCode"].isna()]["currencySymbol"].drop_duplicates()
print("매핑 안 된 통화기호:")
print(unmapped)

# ==========================================
# 2. 환율 API 호출 및 데이터프레임 생성
# ==========================================
# NaN을 제외한 고유한 통화 코드만 리스트로 추출
currency_codes = final_df["currencyCode"].dropna().unique().tolist()
rates_data = []

print("\n환율 정보를 가져오는 중입니다...")
for code in currency_codes:
    # get_avg_rates_to_usd 함수와 날짜 변수들은 위 셀에서 실행해두셨다고 가정합니다.
    rate_1m, rate_6m = get_avg_rates_to_usd(code, start_date_1m, start_date_6m, end_date)
    
    rates_data.append({
        "currencyCode": code,
        "avg_rate_to_usd_1m": rate_1m,
        "avg_rate_to_usd_6m": rate_6m
    })
    
    time.sleep(1) # 차단 방지

rate_df = pd.DataFrame(rates_data)

# ==========================================
# 3. 원본 데이터(final_df)에 환율 데이터 병합(Merge)
# ==========================================
final_df = pd.merge(final_df, rate_df, on="currencyCode", how="left")

# ==========================================
# 4. 특수기호 제거 및 USD 환산 계산
# ==========================================
# 숫자 변환 (이전에 만들어둔 정규표현식 사용)
final_df["campaignGoal"] = pd.to_numeric(final_df["campaignGoal"].astype(str).str.replace(r'[^\d\.]', '', regex=True), errors="coerce")
final_df["fundsGathered"] = pd.to_numeric(final_df["fundsGathered"].astype(str).str.replace(r'[^\d\.]', '', regex=True), errors="coerce")
final_df["price"] = pd.to_numeric(final_df["price"].astype(str).str.replace(r'[^\d\.]', '', regex=True), errors="coerce")
final_df["effectivePrice"] = pd.to_numeric(final_df["effectivePrice"].astype(str).str.replace(r'[^\d\.]', '', regex=True), errors="coerce")


# 1달 기준 환산
final_df["campaignGoal_usd_1m"] = final_df["campaignGoal"] * final_df["avg_rate_to_usd_1m"]
final_df["fundsGathered_usd_1m"] = final_df["fundsGathered"] * final_df["avg_rate_to_usd_1m"]
final_df["price_usd_1m"] = final_df["price"] * final_df["avg_rate_to_usd_1m"]
final_df["effectivePrice"] = final_df["effectivePrice"] * final_df["avg_rate_to_usd_1m"]
# 6달 기준 환산
final_df["campaignGoal_usd_6m"] = final_df["campaignGoal"] * final_df["avg_rate_to_usd_6m"]
final_df["fundsGathered_usd_6m"] = final_df["fundsGathered"] * final_df["avg_rate_to_usd_6m"]
final_df["price_usd_6m"] = final_df["price"] * final_df["avg_rate_to_usd_6m"]
final_df["effectivePrice"] = final_df["effectivePrice"] * final_df["avg_rate_to_usd_6m"]
```

### 이미지 RGB/ks_color/emotion 분리

```python
voca_df
```

```python
voca_df['projectID'] = voca_df['file_name'].str.replace('.jpg', '', case=False)


voca_group_df = voca_df.groupby('projectID')['emotion_adjective'].apply(list).reset_index()
```

```python
import pandas as pd


def parse_rgb(rgb_str):
    clean = rgb_str.replace('(', '').replace(')', '').strip()
    return [int(p) for p in clean.split()]

rgb_parsed = voca_df['RGB'].apply(parse_rgb)
voca_df['R'] = rgb_parsed.apply(lambda x: x[0])
voca_df['G'] = rgb_parsed.apply(lambda x: x[1])
voca_df['B'] = rgb_parsed.apply(lambda x: x[2])

voca_df['idx'] = voca_df.groupby('projectID').cumcount() + 1

pivot_rgb = voca_df.pivot(index='projectID', columns='idx', values=['R', 'G', 'B'])
pivot_rgb.columns = [f'{col[0]}{col[1]}' for col in pivot_rgb.columns]  # R1, G1, B1... 형태 변환

pivot_ks = voca_df.pivot(index='projectID', columns='idx', values='ks_color')
pivot_ks.columns = [f'ks_color_{col}' for col in pivot_ks.columns]

pivot_em = voca_df.pivot(index='projectID', columns='idx', values='emotion_adjective')
pivot_em.columns = [f'emotion_adjective_{col}' for col in pivot_em.columns]


final_voca_df = pd.concat([pivot_rgb, pivot_ks, pivot_em], axis=1).reset_index()

rgb_cols = []
for i in range(1, 5):
    rgb_cols.extend([f'R{i}', f'G{i}', f'B{i}'])
other_cols = [f'ks_color_{i}' for i in range(1, 5)] + [f'emotion_adjective_{i}' for i in range(1, 5)]

final_cols = ['projectID'] + rgb_cols + other_cols
final_voca_df = final_voca_df[final_cols]

```

### 코멘트 그룹 분리(0:upcomming 1:start~+3)

```python
import pandas as pd

sum_cols = ['creator_id', 'is_pledge_master', 'is_backer', 'is_prior_backer', 'is_pathfinder','is_normal','likes']
valid_categories = ['Product_Question', 'Suggestion_Idea', 'Praise_Support', 'Shipping_Fulfillment', 'Complaint_Refund', 'Spam_Irrelevant']


df_sums_group = comment_df.groupby(['projectID', 'group'])[sum_cols].sum().reset_index()
df_cat_group = comment_df.groupby(['projectID', 'group', 'category']).size().unstack(fill_value=0)
df_cat_group = df_cat_group.reindex(columns=valid_categories, fill_value=0).reset_index()


df_merged_group = pd.merge(df_sums_group, df_cat_group, on=['projectID', 'group'], how='outer')
rename_dict = {cat: f"{cat}_count" for cat in valid_categories}
df_merged_group = df_merged_group.rename(columns=rename_dict)


pivot_target_cols = [col for col in df_merged_group.columns if col not in ['projectID', 'group']]
df_pivot = df_merged_group.pivot(index='projectID', columns='group', values=pivot_target_cols)

df_pivot.columns = [f"{col[0]}_{col[1]}" for col in df_pivot.columns]
df_pivot = df_pivot.reset_index()
```
