조인 후 실행

```python
final_df = pd.merge(df1, df2, on='projectID', how='left')
```

```python
df1['campaignStart'] = pd.to_datetime(df1['campaignStart'])
df1['campaignEnd'] = pd.to_datetime(df1['campaignEnd'])


# creatorID -> 그 당시 몇개 만들었는지
final_df = final_df.sort_values(by='campaignStart')
final_df['previous_campaigns_count'] = final_df.groupby('creatorID').cumcount()
final_df = final_df.sort_index()

# 프로젝트 진행 기간 계산
final_df['duration_days'] = (final_df['campaignEnd'] - final_df['campaignStart']).dt.days

# 댓글 수집 기준일 계산 (시작일로부터 1주일, 2주일 뒤) 댓글 조인 후 날짜 잘라도?
#final_df['comment_target_1w'] = final_df['campaignStart'] + pd.Timedelta(days=7)

# playTime 단위 통일 (playTimeUnit이 1이면 시간 단위 -> 분으로 변환)
final_df.loc[final_df['playTimeUnit'] == 1, 'playTime'] = final_df['playTime'] * 60

# pledgeManagerSoftCloseDeadline 존재 여부 (방치 여부/관리 척도)
final_df['softclose'] = final_df['pledgeManagerSoftCloseDeadline'].notna().astype(int)

# projectTags 원핫인코딩
tags_encoded = final_df['projectTags'].fillna('').str.get_dummies(sep=', ')
final_df = pd.concat([final_df, tags_encoded], axis=1)

# final_df = final_df.drop(columns=['playTimeUnit', 'pledgeManagerSoftCloseDeadline', 'projectTags'])
```
