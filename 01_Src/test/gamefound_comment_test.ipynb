```python
import requests
import re
import pandas as pd

# 여러 프로젝트의 댓글 페이지 URL 리스트 (예시)
project_urls = [
    "https://gamefound.com/en/projects/oomm-games/stonesaga-second-printing/comments",
    # "https://gamefound.com/en/projects/다른-제작자/다른-프로젝트/comments" 
]

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
}


for url in project_urls:
    print(f"[{url}] 데이터 수집 중...")
    
    # 1. 프로젝트 댓글 페이지 HTML 가져오기
    res = requests.get(url, headers=headers)
    if res.status_code != 200:
        print(f"페이지 로드 실패: {res.status_code}")
        continue
        
    # 2. 정규표현식으로 HTML 내부에서 commentThreadID 추출
    # 보통 "commentThreadID": 12345 또는 commentThreadId: 12345 형태로 존재합니다.
    match = re.search(r'"commentThreadID"\s*:\s*(\d+)', res.text, re.IGNORECASE)
    
    if match:
        comment_thread_id = int(match.group(1))
        print(f"  👉 commentThreadID 확인됨: {comment_thread_id}")
        
        # 3. 추출한 ID를 페이로드에 삽입
        payload = {
            "GetCommentsForCurrentUser": False,
            "commentID": None,
            "commentThreadID": comment_thread_id, # 여기에 동적으로 꽂아줍니다!
            "freshCommentID": None,
            "getCommentsWithCreatorInput": False,
            "highlightedCommentID": None,
            "lastFetchedCommentID": None,
            "lastPinnedAt": None,
            "lastScore": None,
            "selectedCommentFilter": None,
            "sortType": 0,  
            "tag": None
        }
        
        # 4. API로 댓글 데이터 요청
        api_res = requests.post(api_url, headers=headers, json=payload)
        
        if api_res.status_code == 200:
            data = api_res.json()
            # 예시: 가져온 데이터 구조에 맞게 댓글 내용 파싱 (실제 JSON 구조 확인 필요)
            # comments = data.get('comments', [])
            # all_comments_data.extend(comments)
            print("  ✅ 댓글 데이터 API 호출 성공!")
        else:
            print(f"  ❌ API 호출 실패: {api_res.status_code}")
            
    else:
        print("  ❌ 해당 페이지에서 commentThreadID를 찾을 수 없습니다.")

# 이후 all_comments_data 를 DataFrame으로 변환
# df = pd.DataFrame(all_comments_data)
# print(df.head())
```
