```python
import requests
from bs4 import BeautifulSoup
import re
import json

url = "https://gamefound.com/en/projects/stronghold-games/terraforming-mars-the-legacy-of--mars?ref=homepage-featured-all_1" 
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')


script_tag = soup.find('script', string=re.compile(r'window\.__INITIAL_STATE__'))
```

```python
script_text = script_tag.string
```

```python
script_text
```

```python
json_match = re.search(r'window\.__INITIAL_STATE__\s*=\s*({.*?});\s*$', script_text, re.DOTALL)
json_match
```

```python
json_str = json_match.group(1)
json_str
```

```python
data = json.loads(json_str)
data
```

```python
data['projectState']['statistics']
```
