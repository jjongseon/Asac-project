### 전체 리스트 크롤링

```python
import requests
import pandas as pd
import time
from tqdm.notebook import tqdm

api_url = "https://gamefound.com/api/projectSearch/searchProjectsForCards"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

payload = {
    "creatorID": None,
    "creatorName": None,
    "sortType": 0,
    "term": "",
    "projectPhaseSearchTypes": [],
    "projectBenefits": [],
    "projectTags": [],
    "projectCatalogCategories": [11], #보드게임 카테고리 11번
    "playerAges": [],
    "playerCounts": [],
    "playTimes": [],
    "creator": {"creatorID": None, "name": None},
    "userCommunitySearchTypes": [],
    "source": 5,
    "pageIndex": 0 # 0페이지부터 시작 -> 이거 돌려야함
}

response = requests.post(api_url, headers=headers, json=payload)

# 응답, 전체 페이지 확인
if response.status_code == 200:
    data = response.json()
    total_page =data.get('projects').get('totalPageCount')
    print(f"전체 페이지 : {total_page}")

else:
    print(f"실패 : {response.status_code}")

print("----------------------------------------------")
# 전체페이지 for문 ㄱㄱ / 총 108페이지

game_list = []

for page in tqdm(range(total_page)):

    payload['pageIndex'] = page

    res = requests.post(api_url, headers=headers, json=payload)

    if res.status_code == 200:
        page_data = res.json()
        projects = page_data.get('projects').get('pagedItems')
        
        projects_data = page_data.get('projects')
        first_num = projects_data.get('firstItemNumber')
        last_num = projects_data.get('lastItemNumber')
        print(f"{page}페이지 {first_num}~{last_num}개 진행중")
 
        for item in projects:
            game_dict = {
                'name': item.get('name'),                       # 프로젝트명
                'projectID' :item.get('projectID'),
                'creator': item.get('creator').get('name'),     # 크리에이터 이름
                'creatorID' : item.get('creator').get('creatorID'), # 크리에이터 고유번호
                'homeUrl' : item.get('creator').get('homeUrl'), # 크리에이터 홈페이지 (여기서 몇개 프로젝트 진행한지 확인 가능)
                'currencySymbol': item.get('currencySymbol'),   # 화폐단위  ->  여러개임.....;;
                'campaignGoal': item.get('campaignGoal'),       # 목표 금액
                'fundsGathered': item.get('fundsGathered'),     # 모금액
                'backersCount': item.get('backersCount'),       # 후원자 수
                'campaignStart': item.get('campaignStart'),     # 시작일
                'campaignEnd': item.get('campaignEnd'),         # 마감일
                'description': item.get('shortDescription'),    # 설명
                'project_url': item.get('url'),                  # 프로젝트 링크 -> 이거로 추가 크롤링 해야함
            


                'phaseLabel' : item.get('phaseLabel'),                
                'enableBoardGameProperties': item.get('projectProperties').get('enableBoardGameProperties'), # ?
                'minPlayers': item.get('projectProperties').get('minPlayers'), # 최소인원
                'maxPlayers': item.get('projectProperties').get('maxPlayers'), # 최대인원
                'minAge': item.get('projectProperties').get('minAge'), # 연령제한
                'playTime': item.get('projectProperties').get('playTime'), # 플레이타임
                'playTimeDescription': item.get('projectProperties').get('playTimeDescription'), # 설명? 다 비어 있는거 같은데
                'playTimeUnit': item.get('projectProperties').get('playTimeUnit'), # ?

                'pledgeManagerAvailability': item.get('pledgeManagerAvailability'),
                'fundedInSeconds': item.get('fundedInSeconds'),
                'imageUrl': item.get('imageUrl'),
                'pledgeManagerSoftCloseDeadline': item.get('pledgeManagerSoftCloseDeadline'),
                'projectTags': ", ".join([tag.get('name') for tag in item.get('projectTags')]) if item.get('projectTags') else None,
                'enableShippingOnlyMode': item.get('enableShippingOnlyMode'),
                'originalType': item.get('originalType'),
                        
            }
            game_list.append(game_dict)

    else:
        print(f"{page}페이지 실패")
        pass
    
    time.sleep(1.5)

df = pd.DataFrame(game_list)
```

### 이미지 데이터 추출

```python
import os
import cv2
import numpy as np
import pandas as pd

# 설정
IMAGE_FOLDER = "../../02_Data/main_image"
OUTPUT_CSV = '../../02_Data/processed/image_full_analysis.csv'
VALID_EXTENSIONS = ('.png', '.jpg', '.jpeg', '.bmp', '.tiff')

def get_edge_density(gray_img):
    """엣지 밀도를 계산하는 함수 (Canny 사용)"""
    edges = cv2.Canny(gray_img, 100, 200)
    return np.sum(edges > 0) / edges.size

def analyze_all(image_folder):
    results = []
    
    files = [f for f in os.listdir(image_folder) if f.lower().endswith(VALID_EXTENSIONS)]
    
    for file_name in files:
        full_path = os.path.join(image_folder, file_name)
        img = cv2.imread(full_path)
        if img is None:
            continue
            
        # 색공간 변환
        hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        
        # 지표 계산
        results.append({
            'file_name': os.path.splitext(file_name)[0],
            'edge_density': get_edge_density(gray),
            'saturation': np.mean(hsv[:, :, 1]),
            'brightness': np.mean(hsv[:, :, 2]),
            'contrast': np.std(gray)
        })
        
    return pd.DataFrame(results)

# 실행 및 저장
if os.path.exists(IMAGE_FOLDER):
    df = analyze_all(IMAGE_FOLDER)
    df.to_csv(OUTPUT_CSV, index=False)
    print(f"분석 완료: 총 {len(df)}개의 이미지 파일 처리됨.")
    print(df.head())
```
