```python
import requests
import pandas as pd
import time

api_url = "https://gamefound.com/api/projectSearch/searchProjectsForCards"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

payload = {
    "creatorID": None,
    "creatorName": None,
    "sortType": 0,
    "term": "",
    "projectPhaseSearchTypes": [],
    "projectBenefits": [],
    "projectTags": [],
    "projectCatalogCategories": [11], #보드게임 카테고리 11번
    "playerAges": [],
    "playerCounts": [],
    "playTimes": [],
    "creator": {"creatorID": None, "name": None},
    "userCommunitySearchTypes": [],
    "source": 5,
    "pageIndex": 0 # 0페이지부터 시작 -> 이거 돌려야함
}

response = requests.post(api_url, headers=headers, json=payload)

# 응답, 전체 페이지 확인
if response.status_code == 200:
    data = response.json()
    total_page =data.get('projects').get('totalPageCount')
    print(f"전체 페이지 : {total_page}")
    print("성공")

else:
    print(f"실패 : {response.status_code}")

print("----------------------------------------------")
# 전체페이지 for문 ㄱㄱ / 총 108페이지

game_list = []

for page in range(total_page):
    print(f"{page} 페이지 시작")

    payload['pageIndex'] = page

    res = requests.post(api_url, headers=headers, json=payload)

    if res.status_code == 200:
        page_data = res.json()
        projects = page_data.get('projects').get('pagedItems')
        
        projects_data = page_data.get('projects')
        first_num = projects_data.get('firstItemNumber')
        last_num = projects_data.get('lastItemNumber')
        print(f"{first_num}~{last_num}개 진행중")
 
        for item in projects:
            game_dict = {
                'name': item.get('name'),                       # 프로젝트명
                'creator': item.get('creator').get('name'),     # 크리에이터 이름
                'currencySymbol': item.get('currencySymbol'),   # 화폐단위  ->  여러개임.....;;
                'campaignGoal': item.get('campaignGoal'),       # 목표 금액
                'fundsGathered': item.get('fundsGathered'),     # 모금액
                'backersCount': item.get('backersCount'),       # 후원자 수
                'campaignStart': item.get('campaignStart'),     # 시작일
                'campaignEnd': item.get('campaignEnd'),         # 마감일
                'description': item.get('shortDescription'),    # 설명
                'project_url': item.get('url'),                  # 프로젝트 링크 -> 이거로 추가 크롤링 해야함
                
                'enableBoardGameProperties': item.get('projectProperties').get('enableBoardGameProperties'), # ?
                'minPlayers': item.get('projectProperties').get('minPlayers'), # 최소인원
                'maxPlayers': item.get('projectProperties').get('maxPlayers'), # 최대인원
                'minAge': item.get('projectProperties').get('minAge'), # 연령제한
                'playTime': item.get('projectProperties').get('playTime'), # 플레이타임
                'playTimeDescription': item.get('projectProperties').get('playTimeDescription'), # 설명? 다 비어 있는거 같은데
                'playTimeUnit': item.get('projectProperties').get('playTimeUnit') # ?
                        
            }
            game_list.append(game_dict)

    else:
        print(f"{page}페이지 실패")
        pass
    
    time.sleep(1.5)

df = pd.DataFrame(game_list)
```

```python
import requests
import re

url = "https://gamefound.com/en/projects/oomm-games/stonesaga-second-printing"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36'
}

print("페이지 요청 중...")
response = requests.get(url, headers=headers)
response.raise_for_status()

# 💡 정규식 패턴 설명:
# https://imgcdn.gamefound.com/richtextimage/richtext/ 로 시작하고,
# 그 뒤에 영문자, 숫자, 하이픈(-)이 1개 이상 오며,
# 마지막이 .png, .jpg, .jpeg, .gif, .webp 중 하나로 끝나는 문자열을 모두 찾습니다.
image_pattern = r'https://imgcdn\.gamefound\.com/richtextimage/richtext/[a-zA-Z0-9-]+\.(?:png|jpg|jpeg|gif|webp)'

# re.findall()을 사용하면 패턴과 일치하는 모든 문자열을 리스트 형태로 반환합니다.
extracted_images = re.findall(image_pattern, response.text)

# 중복된 이미지 URL이 있을 수 있으므로 set을 이용해 중복 제거 후 다시 리스트로 변환
unique_images = list(set(extracted_images))

print(f"총 {len(unique_images)}개의 고유한 이미지를 찾았습니다!\n")

print("수집된 이미지 URL:")
for img_url in unique_images:
    print(img_url)
```
